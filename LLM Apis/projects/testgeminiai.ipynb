```python
from openai import OpenAI
import openai
from dotenv import load_dotenv
import pandas as pd
import os
load_dotenv()
```

```python
final_answer=""

client=OpenAI(api_key=os.getenv("GEMINI_API_KEY"),
              base_url="https://generativelanguage.googleapis.com/v1beta/openai/")

# model_list=client.models.list()
# for model in model_list:
#   my_models=pd.DataFrame(model)
#   print(my_models)
# chat completion API
try:
 response=client.chat.completions.create(
    model="gemma-4-31b-it",
     messages=[
        {"role" :"system" ,"content":"You are a helpful Ai assistant Do not use markdown formatting, bold text, or asterisks in your response"},
        #{"role" :"assistant" ,"content":"Hello ! how may i help you"},
          {"role" :"user" ,"content":"what is ML explain in detail"},
       
    ],
    response_format={"type": "text"},
    stream=True,
    temperature=0.2,
    # max_tokens=20
    # max_completion_tokens=2000,
)
 for event in response:
    #the response is coming in chunks so we cant directly apply split on it
    if event.choices[0].delta.content:
      final_answer+=event.choices[0].delta.content
    print(final_answer)
#in case of in valid,expired or revoked API
except openai.AuthenticationError as e :
  print(f"Invalid API Key or unauthorized organization:{e}")
# Invalid model parameters, missing fields, or malformed JSON
except openai.BadRequestError as e:
  print(f"Bad request sent to OpenAI:{e}")
#APIError exception can catch almost all type of errors related to api ,we can use it in case we want to use only one except block to catch most API errors
except openai.APIError as e:
   print(f"An unexpected OpenAI library error occurred: {e}") 

     

#print(response.choices[0].message.content.split("</thought>")[-1].strip())


```
